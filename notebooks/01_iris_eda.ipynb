# Iris Exploratory Data Analysis

Explore class balance and feature relationships in the built-in Iris dataset. No external data download is needed.

```python
import matplotlib.pyplot as plt
from src.data import load_iris_frame
from src.eda import species_counts, summarize_by_species

iris_df = load_iris_frame()
iris_df.head()
```

## Class balance

The dataset contains the same number of observations for each species.

```python
species_counts(iris_df)
```

## Feature summary

Compare the mean and standard deviation of each measurement by species.

```python
summary_by_species = summarize_by_species(iris_df)
summary_by_species
```

## Feature relationship

Petal length separates the species more clearly than sepal length in this two-feature view.

```python
figure, axis = plt.subplots(figsize=(7, 5))
for species, group in iris_df.groupby("species", sort=True):
    axis.scatter(
        group["sepal length (cm)"],
        group["petal length (cm)"],
        label=species,
    )
axis.set(title="Iris measurements by species", xlabel="Sepal length (cm)", ylabel="Petal length (cm)")
axis.legend(title="Species")
figure.tight_layout()
```
