# Iris Baseline Classifier

Train a scaled logistic-regression baseline using a stratified 80/20 split and a fixed random seed.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

from src.data import load_iris_frame
from src.model import evaluate_model, train_baseline

iris_df = load_iris_frame()
model, X_test, y_test = train_baseline(iris_df)
metrics = evaluate_model(model, X_test, y_test)
metrics
```

## Hold-out performance

The test split is stratified to preserve the species distribution. Accuracy and balanced accuracy are reported together.

```python
figure, axis = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_estimator(
    model,
    X_test,
    y_test,
    display_labels=model.classes_,
    xticks_rotation="vertical",
    ax=axis,
)
axis.set_title("Iris baseline confusion matrix")
figure.tight_layout()
```

## Interpretation

The confusion matrix reveals which species are misclassified. This small toy dataset is suitable for learning and should not be treated as a production performance estimate.
